```python
import os
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class BadmintonDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        clip_path = self.paths[idx]
        frames = []

        for i in range(16):
            image_path = os.path.join(
                clip_path, f"frame_{i:02d}.jpg"
            )

            image = Image.open(image_path).convert("RGB")

            if self.transform is not None:
                image = self.transform(image)

            frames.append(image)

        video = torch.stack(frames)
        label = int(self.labels[idx])

        return video, label

print("Dataset class created successfully!")
```

```python
from torch.utils.data import DataLoader

train_dataset = BadmintonDataset(
    train_paths, train_labels, transform
)

val_dataset = BadmintonDataset(
    val_paths, val_labels, transform
)

test_dataset = BadmintonDataset(
    test_paths, test_labels, transform
)

batch_size = 4

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=2
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=2
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=2
)

print("Training clips:", len(train_dataset))
print("Validation clips:", len(val_dataset))
print("Testing clips:", len(test_dataset))
print("DataLoaders created successfully!")
```

```python
import torch
import torch.nn as nn
from torchvision.models import resnet18, ResNet18_Weights

class CNNLSTMPretrained(nn.Module):
    def __init__(self, num_classes=5):
        super().__init__()

        # CNN: extract features from each frame
        self.cnn = resnet18(weights=ResNet18_Weights.DEFAULT)
        self.cnn.fc = nn.Identity()

        # LSTM: learn movement across the frames
        self.lstm = nn.LSTM(
            input_size=512,
            hidden_size=256,
            num_layers=2,
            batch_first=True,
            dropout=0.3
        )

        # Final classification layers
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(256, num_classes)

    def forward(self, x):
        batch_size, sequence_length, C, H, W = x.shape

        # Process each frame through the CNN
        x = x.reshape(batch_size * sequence_length, C, H, W)
        x = self.cnn(x)

        # Restore the sequence structure
        x = x.reshape(batch_size, sequence_length, 512)

        # Learn temporal patterns
        x, _ = self.lstm(x)

        # Use the final timestep for classification
        x = x[:, -1, :]
        x = self.dropout(x)

        return self.fc(x)

print("CNN-LSTM model class created successfully!")
```

```python
import torch
import torch.nn as nn

# Initialize the complete CNN-LSTM model
model = CNNLSTMPretrained(num_classes=5).to(device)

# Optimizer updates the model's learnable parameters
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.0001,
    weight_decay=0.0001
)

# Weighted loss gives extra importance to underrepresented classes
criterion = nn.CrossEntropyLoss(weight=class_weights)

print("Model initialized successfully!")
print("Device:", device)
print("LSTM layers:", model.lstm.num_layers)
print("Hidden size:", model.lstm.hidden_size)
print("Output classes:", model.fc.out_features)
print("Optimizer: AdamW")
print("Learning rate: 0.0001")
print("Loss: Weighted CrossEntropyLoss")
```

```python
import os
import torch
from tqdm.auto import tqdm

epochs = 10
best_val_acc = 0.0

checkpoint_path = (
    "/content/drive/MyDrive/"
    "badminton_pretrained_2lstm256_best.pth"
)

for epoch in range(epochs):
    # ---- Training ----
    model.train()
    train_correct = 0
    train_total = 0
    train_loss_sum = 0.0

    train_bar = tqdm(train_loader, desc=f"Epoch {epoch + 1}/{epochs} [Train]")

    for videos, labels in train_bar:
        videos = videos.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(videos)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        train_loss_sum += loss.item() * labels.size(0)
        train_correct += (outputs.argmax(dim=1) == labels).sum().item()
        train_total += labels.size(0)

        train_bar.set_postfix(loss=f"{loss.item():.4f}")

    train_loss = train_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # ---- Validation ----
    model.eval()
    val_correct = 0
    val_total = 0
    val_loss_sum = 0.0

    with torch.no_grad():
        for videos, labels in tqdm(
            val_loader, desc=f"Epoch {epoch + 1}/{epochs} [Val]"
        ):
            videos = videos.to(device)
            labels = labels.to(device)

            outputs = model(videos)
            loss = criterion(outputs, labels)

            val_loss_sum += loss.item() * labels.size(0)
            val_correct += (outputs.argmax(dim=1) == labels).sum().item()
            val_total += labels.size(0)

    val_loss = val_loss_sum / val_total
    val_acc = 100.0 * val_correct / val_total

    print(
        f"\nEpoch {epoch + 1}/{epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.2f}%"
    )

    # Save the best validation checkpoint
    if val_acc > best_val_acc:
        best_val_acc = val_acc

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "best_val_acc": best_val_acc,
                "class_names": class_names,
            },
            checkpoint_path,
        )

        print(f"Best checkpoint saved: {checkpoint_path}")

print(f"\nTraining complete. Best validation accuracy: {best_val_acc:.2f}%")
```

```python
from pathlib import Path
from sklearn.model_selection import GroupShuffleSplit
from collections import Counter

# Collect every clip and its existing label.
datasets = [train_dataset, val_dataset, test_dataset]

all_records = []
for ds in datasets:
    all_records.extend(zip(ds.paths, ds.labels))

# Check for duplicate clip paths.
path_to_label = {}
for path, label in all_records:
    if path in path_to_label and path_to_label[path] != label:
        raise ValueError(f"Conflicting labels for {path}")
    path_to_label[path] = label

all_records = list(path_to_label.items())
paths = [p for p, y in all_records]
labels = [int(y) for p, y in all_records]

print("Unique clips:", len(paths))

# Provisional grouping rule:
# FB: group all events from the same FB recording.
# VB: group clips sharing timestamp and camera identifiers.
def recording_group(path):
    name = Path(path).name
    parts = name.split("_")

    if parts[0].startswith("FB") and len(parts) >= 4:
        return "_".join(parts[:2])

    if parts[0] == "VB" and len(parts) >= 12:
        return "_".join(parts[3:5] + parts[10:12])

    return name

groups = [recording_group(p) for p in paths]

# Split off 10% test groups.
split1 = GroupShuffleSplit(n_splits=1, test_size=0.10, random_state=42)
trainval_idx, test_idx = next(
    split1.split(paths, labels, groups)
)

# Split 10% of the complete dataset for validation.
# 0.10 / 0.90 = 1/9 of the remaining data.
remaining_groups = [groups[i] for i in trainval_idx]
split2 = GroupShuffleSplit(n_splits=1, test_size=1/9, random_state=43)

train_rel, val_rel = next(
    split2.split(
        [paths[i] for i in trainval_idx],
        [labels[i] for i in trainval_idx],
        remaining_groups
    )
)

train_idx = [trainval_idx[i] for i in train_rel]
val_idx = [trainval_idx[i] for i in val_rel]

train_paths_new = [paths[i] for i in train_idx]
train_labels_new = [labels[i] for i in train_idx]

val_paths_new = [paths[i] for i in val_idx]
val_labels_new = [labels[i] for i in val_idx]

test_paths_new = [paths[i] for i in test_idx]
test_labels_new = [labels[i] for i in test_idx]

# Verify group separation.
train_groups_new = {groups[i] for i in train_idx}
val_groups_new = {groups[i] for i in val_idx}
test_groups_new = {groups[i] for i in test_idx}

print("\nNew split sizes:")
print("Train:", len(train_paths_new))
print("Validation:", len(val_paths_new))
print("Test:", len(test_paths_new))

print("\nGroup overlap:")
print("Train vs validation:", len(train_groups_new & val_groups_new))
print("Train vs test:", len(train_groups_new & test_groups_new))
print("Validation vs test:", len(val_groups_new & test_groups_new))

print("\nClass counts (numeric labels):")
print("Train:", Counter(train_labels_new))
print("Validation:", Counter(val_labels_new))
print("Test:", Counter(test_labels_new))
```

```python
import torch
import torch.nn as nn
from torch.optim import AdamW
from tqdm.auto import tqdm

# Training configuration
epochs = 5
learning_rate = 2e-5
weight_decay = 1e-4

criterion = nn.CrossEntropyLoss()

optimizer = AdamW(
    model.parameters(),
    lr=learning_rate,
    weight_decay=weight_decay
)

best_val_acc = 0.0

for epoch in range(epochs):
    model.train()
    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(train_loader_new, desc=f"Epoch {epoch+1}"):
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * labels.size(0)
        train_correct += (outputs.argmax(1) == labels).sum().item()
        train_total += labels.size(0)

    train_loss /= train_total
    train_acc = 100 * train_correct / train_total

    model.eval()
    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader_new:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_loss += loss.item() * labels.size(0)
            val_correct += (outputs.argmax(1) == labels).sum().item()
            val_total += labels.size(0)

    val_loss /= val_total
    val_acc = 100 * val_correct / val_total

    print(
        f"\nEpoch {epoch+1}/{epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.2f}%"
    )

    if val_acc > best_val_acc:
        best_val_acc = val_acc

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "best_val_acc": best_val_acc,
            },
            "/content/drive/MyDrive/"
            "badminton_group_split_best.pth",
        )

        print("Best new-split checkpoint saved!")

print(f"\nBest validation accuracy: {best_val_acc:.2f}%")
```
